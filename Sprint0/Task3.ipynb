<h2 style="color:DodgerBlue">Описание проекта:</h2>Создать базовый класс BankAccount в C#, который будет представлять информацию об учётных записях в банке. На основе этого класса разработать 2–3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

```polyglot-notebook
using System;
using System.Collections.Generic;

class BankAccount
{
    public string Owner { get; set; }
    public decimal Balance { get; protected set; }

    public BankAccount(string owner, decimal balance)
    {
        Owner = owner;
        Balance = balance;
    }

    public virtual void Withdraw(decimal amount)
    {
        if (amount > Balance) { Console.WriteLine("Недостаточно средств"); return; }
        Balance -= amount;
        Console.WriteLine($"Снято {amount:C}. Баланс: {Balance:C}");
    }

    public virtual string GetInfo() => $"{Owner} | Баланс: {Balance:C}";
}

class SavingsAccount : BankAccount
{
    public decimal MinBalance { get; set; } = 100m;
    public SavingsAccount(string owner, decimal balance) : base(owner, balance) { }

    public void ApplyInterest()
    {
        Balance *= 1.05m;
        Console.WriteLine($"Проценты начислены. Баланс: {Balance:C}");
    }

    public override void Withdraw(decimal amount)
    {
        if (Balance - amount < MinBalance) { Console.WriteLine($"Нельзя: минимум {MinBalance:C}"); return; }
        base.Withdraw(amount);
    }

    public override string GetInfo() => base.GetInfo() + " | Сберегательный";
}

class CreditAccount : BankAccount
{
    public decimal Limit { get; set; } = 5000m;
    public CreditAccount(string owner, decimal balance) : base(owner, balance) { }

    public void ShowDebt() => Console.WriteLine(Balance < 0 ? $"Долг: {-Balance:C}" : "Долга нет");

    public override void Withdraw(decimal amount)
    {
        if (Balance - amount < -Limit) { Console.WriteLine("Превышен лимит"); return; }
        Balance -= amount;
        Console.WriteLine($"Снято {amount:C}. Баланс: {Balance:C}");
    }

    public override string GetInfo() => base.GetInfo() + " | Кредитный";
}

class DepositAccount : BankAccount
{
    public DepositAccount(string owner, decimal balance) : base(owner, balance) { }

    public override void Withdraw(decimal amount)
    {
        Console.WriteLine("Досрочно: штраф 10%");
        base.Withdraw(amount * 1.1m);
    }

    public override string GetInfo() => base.GetInfo() + " | Вклад";
}
```

```polyglot-notebook
List<BankAccount> accounts = new()
{
    new SavingsAccount("Иван", 1000m),
    new CreditAccount("Мария", 500m),
    new DepositAccount("Олег", 2000m)
};

foreach (var acc in accounts)
    Console.WriteLine(acc.GetInfo());

Console.WriteLine("\n--- Операции ---");
accounts[0].Withdraw(950m);
accounts[1].Withdraw(3000m);
accounts[2].Withdraw(500m);

Console.WriteLine("\n--- Доп. методы ---");
((SavingsAccount)accounts[0]).ApplyInterest();
((CreditAccount)accounts[1]).ShowDebt();
```
